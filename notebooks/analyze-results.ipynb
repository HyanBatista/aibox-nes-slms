## Import Libraries

```python
import json
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns

# Set display options
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)
```

## Load Evaluation Metrics Function

This function reads all evaluation_metrics.json files from the processed data directory and organizes them into a pandas DataFrame.

```python
def load_evaluation_metrics(processed_data_path="../data/processed"):
    """
    Load evaluation metrics from all model directories and organize into a DataFrame.
    
    Args:
        processed_data_path: Path to the processed data directory
        
    Returns:
        DataFrame with columns: model, formal_register, thematic_coherence,
        narrative_rhetorical_structure, cohesion, and overall metrics
    """
    processed_dir = Path(processed_data_path)
    
    # List to store all metrics
    metrics_list = []
    
    # Iterate through all subdirectories
    for model_dir in processed_dir.iterdir():
        if model_dir.is_dir():
            metrics_file = model_dir / "evaluation_metrics.json"
            
            if metrics_file.exists():
                # Load the JSON file
                with open(metrics_file, 'r') as f:
                    data = json.load(f)
                
                # Extract the model name (folder name)
                model_name = model_dir.name
                
                # Extract competency metrics
                competencies = data.get('competencies', {})
                
                # Create a row for this model
                row = {
                    'model': model_name,
                    # Formal Register metrics
                    'formal_register_accuracy': competencies.get('formal_register', {}).get('accuracy'),
                    'formal_register_precision': competencies.get('formal_register', {}).get('precision'),
                    'formal_register_recall': competencies.get('formal_register', {}).get('recall'),
                    'formal_register_f1': competencies.get('formal_register', {}).get('f1_score'),
                    'formal_register_exact_match': competencies.get('formal_register', {}).get('exact_match_rate'),
                    'formal_register_off_by_one': competencies.get('formal_register', {}).get('off_by_one_rate'),
                    # Thematic Coherence metrics
                    'thematic_coherence_accuracy': competencies.get('thematic_coherence', {}).get('accuracy'),
                    'thematic_coherence_precision': competencies.get('thematic_coherence', {}).get('precision'),
                    'thematic_coherence_recall': competencies.get('thematic_coherence', {}).get('recall'),
                    'thematic_coherence_f1': competencies.get('thematic_coherence', {}).get('f1_score'),
                    'thematic_coherence_exact_match': competencies.get('thematic_coherence', {}).get('exact_match_rate'),
                    'thematic_coherence_off_by_one': competencies.get('thematic_coherence', {}).get('off_by_one_rate'),
                    # Narrative Rhetorical Structure metrics
                    'narrative_rhetorical_structure_accuracy': competencies.get('narrative_rhetorical_structure', {}).get('accuracy'),
                    'narrative_rhetorical_structure_precision': competencies.get('narrative_rhetorical_structure', {}).get('precision'),
                    'narrative_rhetorical_structure_recall': competencies.get('narrative_rhetorical_structure', {}).get('recall'),
                    'narrative_rhetorical_structure_f1': competencies.get('narrative_rhetorical_structure', {}).get('f1_score'),
                    'narrative_rhetorical_structure_exact_match': competencies.get('narrative_rhetorical_structure', {}).get('exact_match_rate'),
                    'narrative_rhetorical_structure_off_by_one': competencies.get('narrative_rhetorical_structure', {}).get('off_by_one_rate'),
                    # Cohesion metrics
                    'cohesion_accuracy': competencies.get('cohesion', {}).get('accuracy'),
                    'cohesion_precision': competencies.get('cohesion', {}).get('precision'),
                    'cohesion_recall': competencies.get('cohesion', {}).get('recall'),
                    'cohesion_f1': competencies.get('cohesion', {}).get('f1_score'),
                    'cohesion_exact_match': competencies.get('cohesion', {}).get('exact_match_rate'),
                    'cohesion_off_by_one': competencies.get('cohesion', {}).get('off_by_one_rate'),
                    # Overall metrics
                    'overall_accuracy': data.get('overall', {}).get('accuracy'),
                    'overall_precision': data.get('overall', {}).get('precision'),
                    'overall_recall': data.get('overall', {}).get('recall'),
                    'overall_f1': data.get('overall', {}).get('f1_score'),
                    'overall_exact_match': data.get('overall', {}).get('exact_match_rate'),
                    'overall_off_by_one': data.get('overall', {}).get('off_by_one_rate'),
                    # Additional info
                    'total_samples': data.get('total_samples'),
                    'valid_samples': data.get('valid_samples'),
                    'parsing_errors': data.get('parsing_errors'),
                }
                
                metrics_list.append(row)
    
    # Create DataFrame
    df = pd.DataFrame(metrics_list)
    
    # Sort by model name for consistency
    df = df.sort_values('model').reset_index(drop=True)
    
    return df
```

## Load All Metrics

```python
# Load all evaluation metrics
df_full = load_evaluation_metrics()

print(f"Found {len(df_full)} models with evaluation metrics:")
print(df_full['model'].tolist())
```
